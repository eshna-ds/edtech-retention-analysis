```python
import pandas as pd
```

```python
file_path =r'C:\Users\HP\projects\EdTech Retention Analytics\data\processed\weekly_activity.csv'
df = pd.read_csv(file_path)
```

```python
# Login risk

def login_risk(total_logins):
    if total_logins <= 1:
        return 100
    elif total_logins <= 3:
        return 60
    else:
        return 20
```

```python
# Assignment risk

def assignment_risk(total_assignments):
    if total_assignments == 0:
        return 100
    elif total_assignments == 1:
        return 60
    else:
        return 20
```

```python
# Apply risk scores

df['login_risk'] = df['total_logins'].apply(login_risk)
df['assignment_risk'] = df['total_assignments'].apply(assignment_risk)
```

```python
# Final risk score

df['final_risk_score'] = (
    0.5 * df['login_risk'] +
    0.5 * df['assignment_risk']
)
```

```python
# Risk level

def risk_level(score):
    if score >= 80:
        return 'High Risk'
    elif score >= 50:
        return 'Medium Risk'
    else:
        return 'Low Risk'
```

```python
df['risk_level'] = df['final_risk_score'].apply(risk_level)

df.head()
```

```python
# Save results

df.to_csv(r'C:\Users\HP\projects\EdTech Retention Analytics\/data/processed/risk_scores.csv', index=False)
```

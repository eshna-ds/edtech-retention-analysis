# DATA CLEANING

```python
import pandas as pd
```

```python
file_path = r'C:\Users\HP\projects\EdTech Retention Analytics\data\raw\mooc_data.csv'
df = pd.read_csv(file_path)
```

```python
print("Dataset Loaded Successfully")
df.head()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Handle missing values
print(df.isnull().sum())
```

```python
# Rename columns
df = df.rename(columns={
    'userid_DI': 'student_id',
    'ndays_act': 'login',
    'nplay_video': 'video_watched',
    'grade': 'assignment_submitted',
    'nforum_posts': 'forum_activity',
    'nevents': 'engagement'
})

# Create dropout column
df['dropout'] = 1 - df['certified']
```

```python
df.head()
```

```python
print(df.shape)
```

```python
# Fill missing numerical values with 0
numeric_columns = ['login', 'video_watched', 'assignment_submitted']
for col in numeric_columns:
    df[col] = df[col].fillna(0)
```

```python
# Convert activity date into datetime
df['last_event_DI'] = pd.to_datetime(df['last_event_DI'])
```

```python
# Create week column
df['week'] = df['last_event_DI'].dt.isocalendar().week
```

```python
# Save cleaned data
output_path = r'C:\Users\HP\projects\EdTech Retention Analytics\data\processed\cleaned_data.csv'
df.to_csv(output_path, index=False)
```

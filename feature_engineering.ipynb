# FEATURE EXTRACTION

```python
import pandas as pd

# Load cleaned data
file_path = r'C:\Users\HP\projects\EdTech Retention Analytics\data\processed\cleaned_data.csv'
df = pd.read_csv(file_path)
```

```python
# Create student weekly activity summary
weekly_activity = df.groupby(['student_id', 'week']).agg({
    'login': 'sum',
    'video_watched': 'sum',
    'assignment_submitted': 'sum'
}).reset_index()
```

```python
# Rename columns
weekly_activity.columns = [
    'student_id',
    'week',
    'total_logins',
    'total_videos',
    'total_assignments'
]
```

```python
# Create dropout label
weekly_activity['dropout'] = weekly_activity['total_logins'].apply(
    lambda x: 1 if x < 2 else 0)
```

```python
# Save processed dataset
weekly_activity.to_csv(
    r'C:\Users\HP\projects\EdTech Retention Analytics\data\processed\weekly_activity.csv',
    index=False
)
```
